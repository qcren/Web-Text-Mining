**Import Libraries**

```python
# Imports the requests library, which allows you to send HTTP requests.
import requests 

# It is used for parsing HTML and XML documents.
from bs4 import BeautifulSoup
```

```python
# Fetch the Wikipedia page
url = "https://en.wikipedia.org/wiki/Adventure_Time"

# It sends a GET request to the specified URL and stores the response.
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
print(soup)
```

**Parsing**

```python
# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(response.content, "html.parser")
```

**Let's find the Title**

```python
# Extract the page title
title = soup.find("h1").text
print("Page Title:", title)
```

**Extract the Introduction Section**

```python
# Extract the introduction section
#intro = soup.find("div", {"class": "mw-parser-output"}).find("p").text
#print("Introduction:", intro)


# Extract Introduction if it exists
intro_div = soup.find("div", class_="mw-parser-output")
if intro_div:
    intro_p = intro_div.find("p")
    if intro_p:
        intro = intro_p.text.strip()
        print("Introduction:", intro)
    else:
        print("Introduction not found")
else:
    print("Introduction div not found")
```

**SPAN - Lets fetch the section History**

```python
# Find the section containing the History content
history_section = soup.find("span", {"id": "History"})

# If the History section exists, get its content
if history_section:
    
    # Get the parent section (usually an <h2> tag)
    history_heading = history_section.find_parent("h2")
    
    # Initialize a list to store content paragraphs
    history_content = []
    
    # Get all sibling elements until the next <h2> tag and extract text
    next_sibling = history_heading.find_next_sibling()
    while next_sibling and next_sibling.name != "h2":
        if next_sibling.name == "p":
            history_content.append(next_sibling.get_text())
        next_sibling = next_sibling.find_next_sibling()
    
    print("History Content:")
    for paragraph in history_content:
        print(paragraph.strip())  # Strip to remove leading and trailing whitespaces
else:
    print("History section not found.")
```

**Extract Main Section Content**

```python
# Extract the key sections of the main content page 
sections = soup.find_all("h2")
for section in sections:
    section_title = section.text
    section_content_tag = section.find_next("p")
    if section_content_tag:
        section_content = section_content_tag.text
        print("Section Title:", section_title)
        print("Section Content:", section_content)
        print()
```

**Extract References**

```python
# Extract the external reference links
references_links = soup.find_all("a", {"class": "external text"})
print("References Links:")
for link in references_links:
    print(link.get("href"))
```

```python
# Import TextBlob for sentiment analysis
from textblob import TextBlob

# Sentiment Analysis Function using TextBlob
def sentiment_analysis(text):
    blob = TextBlob(text)
    return blob.sentiment.polarity, blob.sentiment.subjectivity

# Apply sentiment analysis to each section
# Title
title_polarity, title_subjectivity = sentiment_analysis(title)
print(f"\nSentiment of Title -> Polarity: {title_polarity}, Subjectivity: {title_subjectivity}")


# Introductory Paragraph
if 'intro' in locals() and intro:  # Ensure 'intro' is defined
    intro_polarity, intro_subjectivity = sentiment_analysis(intro)
    print(f"\nSentiment of Introduction -> Polarity: {intro_polarity}, Subjectivity: {intro_subjectivity}")
else:
    print("No Introduction content to analyze.")


# History
if 'history_content' in locals() and history_content:  # Ensure history_content is defined and has content
    history_text = ' '.join(history_content)  # Combine paragraphs into a single string for analysis
    history_polarity, history_subjectivity = sentiment_analysis(history_text)
    print(f"\nSentiment of History Section -> Polarity: {history_polarity}, Subjectivity: {history_subjectivity}")
else:
    print("No History content to analyze.")

# Apply sentiment analysis to the Main Content sections
sections = soup.find_all("h2")
for section in sections:
    section_title = section.text
    section_content_tag = section.find_next("p")
    if section_content_tag:
        section_content = section_content_tag.text.strip()
        section_polarity, section_subjectivity = sentiment_analysis(section_content)
        print(f"Sentiment of Section '{section_title}' -> Polarity: {section_polarity}, Subjectivity: {section_subjectivity}")

# Apply sentiment analysis to the References
print("\nAnalyzing References:")
for ref in references_links[:5]:  # Limiting to the first 5 references for brevity
    reference_url = ref.get("href")
    print(f"Reference URL: {reference_url}")
```

```python
# Extract the external reference links
references_links = soup.find_all("a", {"class": "external text"})
print("References Links Text Sentiment Analysis:")

for link in references_links:
    link_text = link.get_text()  # Extract the text inside the <a> tag
    blob = TextBlob(link_text)   # Perform sentiment analysis
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity
    print(f"Link Text: {link_text}")
    print(f"Polarity: {polarity}, Subjectivity: {subjectivity}")
    print("---")
```
